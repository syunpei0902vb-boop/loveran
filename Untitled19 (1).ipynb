```python
import requests
from bs4 import BeautifulSoup
import json
import os
import re
from datetime import datetime

BASE_URL = "https://www.loveran.shop/view/category/all_items?page={}"
HEADERS = {
    "User-Agent": "Mozilla/5.0"
}

JSON_FILE = "products.json"

# ✅ ちゃんと入れる
DISCORD_WEBHOOK_URL = "https://discord.com/api/webhooks/1508073467352977590/ID0SXkyltKGrBTdmrk5EwxQZTC-0lzfKNhh4NWmQitePc7AvKL-bxecQJXHQLB-GCJtd"

def extract_price(text):
    if not text:
        return None
    text = text.replace(",", "")
    m = re.search(r"\d+", text)
    return int(m.group()) if m else None

def get_products():

    products = {}
    page = 1

    while True:

        print(f"取得中 page={page}")

        url = BASE_URL.format(page)
        r = requests.get(url, headers=HEADERS)

        if r.status_code != 200:
            break

        soup = BeautifulSoup(r.text, "html.parser")

        items = soup.select('a[href*="/view/item/"]')

        if not items:
            break

        count_before = len(products)

        for item in items:

            href = item.get("href")
            if not href:
                continue

            name = item.get_text(strip=True)

            # ✅ 修正ポイント（ここ）
            price_tag = item.find_next(class_="price")
            price = extract_price(price_tag.text) if price_tag else None

            product_url = "https://www.loveran.shop" + href
            product_id = href.split("/")[-1]

            if not name:
                continue

            products[product_id] = {
                "id": product_id,
                "name": name,
                "price": price,
                "url": product_url,
                "updated_at": datetime.now().isoformat()
            }

        if len(products) == count_before:
            break

        page += 1

    return products

def load_old_data():
    if not os.path.exists(JSON_FILE):
        return {}
    with open(JSON_FILE, "r", encoding="utf-8") as f:
        return json.load(f)

def save_data(data):
    with open(JSON_FILE, "w", encoding="utf-8") as f:
        json.dump(data, f, ensure_ascii=False, indent=2)

def compare_products(old, new):

    changes = []

    for k in new.keys() - old.keys():
        changes.append(f"🆕 新商品: {new[k]['name']} ¥{new[k]['price']}")

    for k in old.keys() - new.keys():
        changes.append(f"❌ 削除: {old[k]['name']}")

    for k in old.keys() & new.keys():

        old_price = old[k]["price"]
        new_price = new[k]["price"]

        if old_price != new_price:

            if old_price and new_price:

                if new_price < old_price:
                    changes.append(f"📉 値下げ: {new[k]['name']} {old_price} → {new_price}")
                else:
                    changes.append(f"📈 値上げ: {new[k]['name']} {old_price} → {new_price}")

    return changes

def send_discord(message):

    if not DISCORD_WEBHOOK_URL:
        print("Webhook未設定")
        return

    res = requests.post(DISCORD_WEBHOOK_URL, json={"content": message})

    print("Discord status:", res.status_code, res.text)

def main():

    is_first_run = not os.path.exists(JSON_FILE)

    old_data = load_old_data()
    new_data = get_products()

    save_data(new_data)

    print("商品数:", len(new_data))

    if is_first_run:

        msg = f"✅ 初回取得完了\n商品数: {len(new_data)}件"
        send_discord(msg)
        print(msg)
        return

    changes = compare_products(old_data, new_data)

    if changes:
        msg = "🔔 変更検知\n" + "\n".join(changes[:20])
        send_discord(msg)
        print(msg)
    else:
        print("変更なし")

if __name__ == "__main__":
    main()
    import subprocess

def git_push():

    try:
        subprocess.run(["git", "config", "--global", "user.name", "github-actions"])
        subprocess.run(["git", "config", "--global", "user.email", "actions@github.com"])

        subprocess.run(["git", "add", "products.json"])
        subprocess.run(["git", "commit", "-m", "update data"])
        subprocess.run(["git", "push"])

    except Exception as e:
        print("git error:", e)
```
